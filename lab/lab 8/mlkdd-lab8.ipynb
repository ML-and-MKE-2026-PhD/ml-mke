# Теоретичні відомості

*Наукове твердження (scientific claim)* -- формулювання, яке має визначені предмет, область застосування та умови перевірки. *Свідчення (evidence)* -- результат спостереження, експерименту або аналізу, який за певних припущень підтримує чи спростовує твердження. *Простежуваність (traceability)* -- можливість перейти від висновку до первинного джерела, даних, коду й протоколу [@fayyad1996kdd; @breiman2001two].

Одна корисна формалізація: нехай $C$ -- множина тверджень, $E$ -- множина свідчень, а $R\subseteq C\times E$ -- відношення підтримки. Тоді наявність ребра $(c,e)\in R$ означає *заявлений* зв'язок, який потрібно окремо перевірити за змістом і методологією. Саме існування посилання не доводить істинності $c$.

*Фундаментальна модель (foundation model)* може допомогти сформулювати гіпотезу, знайти фрагмент тексту чи структурувати звіт. Вона не замінює первинного джерела, коректного експерименту або причинної ідентифікації [@bommasani2021foundation; @pearl2009causality].

## 2. Пошук потенційних свідчень: векторні представлення

Створимо синтетичний корпус із шести коротких текстів. TF-IDF є прозорим базовим представленням для демонстрації пошуку; це *не* фундаментальна модель і не семантичний embedding трансформера. Мета -- відокремити процедуру пошуку від процедури перевірки доказів.


````python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

SEED = 42
np.random.seed(SEED)

sources = pd.DataFrame([
    {"source_id":"D01", "text":"Group-aware validation prevents leakage between repeated observations of the same subject.", "type":"synthetic_note"},
    {"source_id":"D02", "text":"SHAP attributes model predictions to features under a chosen reference distribution.", "type":"synthetic_note"},
    {"source_id":"D03", "text":"Predictive accuracy alone does not identify the causal effect of a treatment.", "type":"synthetic_note"},
    {"source_id":"D04", "text":"Cross-fitting separates nuisance model training from evaluation of orthogonal scores.", "type":"synthetic_note"},
    {"source_id":"D05", "text":"Clustering stability is a diagnostic for robustness of discovered partitions.", "type":"synthetic_note"},
    {"source_id":"D06", "text":"A retrieved document may be relevant without proving a scientific claim.", "type":"synthetic_note"},
])
query = "Does predictive model explanation establish a causal treatment effect?"
vectorizer = TfidfVectorizer(lowercase=True, stop_words="english", ngram_range=(1,2))
D = vectorizer.fit_transform(sources["text"])
q = vectorizer.transform([query])
scores = cosine_similarity(q, D).ravel()
ranked = sources.assign(similarity=scores).sort_values("similarity", ascending=False)
display(ranked[["source_id","similarity","text"]].round(3))
````

**Завдання 2.1.** Поясніть, чому найближчий за косинусною подібністю текст може бути недостатнім для підтвердження наукового твердження. Змініть формулювання запиту й порівняйте ранжування.

## 3. Реєстр тверджень і доказів

Для кожного твердження фіксуємо: ідентифікатор, текст, тип (прогнозне, пояснювальне, причинне), джерела, протокол перевірки, статус, обмеження. Статус *supported* дозволений лише після предметної та методологічної перевірки, а не після автоматичного пошуку.

*Завдання 3.1.* Побудуйте реєстр на основі штучного корпусу; свідомо залиште хоча б одне твердження без достатнього підтвердження.


````python
claims = pd.DataFrame([
    {"claim_id":"C01", "claim":"SHAP attribution proves that a feature causes an outcome", "kind":"causal", "source_ids":["D02"], "status":"unsupported", "limitation":"Attribution concerns a model prediction, not an intervention."},
    {"claim_id":"C02", "claim":"Group-aware splitting can prevent subject overlap across train and test", "kind":"validation", "source_ids":["D01"], "status":"illustrated", "limitation":"Requires correct group identifiers and design."},
    {"claim_id":"C03", "claim":"Cross-fitting separates nuisance training from out-of-fold prediction", "kind":"method", "source_ids":["D04"], "status":"illustrated", "limitation":"Does not remove unmeasured confounding."},
    {"claim_id":"C04", "claim":"An untested model generalizes to all new populations", "kind":"external_validity", "source_ids":[], "status":"unsupported", "limitation":"External validation absent."},
])
source_ids = set(sources["source_id"])
claims["references_exist"] = claims["source_ids"].apply(
    lambda ids: bool(ids) and all(sid in source_ids for sid in ids)
)
claims["needs_review"] = (claims["status"] != "supported")
display(claims[["claim_id","kind","source_ids","status","references_exist","needs_review"]])
````

Зверніть увагу: у цьому навчальному прикладі статус *illustrated* означає, що ідею продемонстровано на синтетичній нотатці, *але не підтверджено незалежною науковою публікацією*. Навіть коректне посилання на реальну статтю потребує перевірки контексту, методів, вибірки та обсягу підтримки твердження.

## 4. Аудит простежуваності

Побудуємо двочастковий граф «твердження -- джерело» та виявимо відсутні посилання. Це перевірка *структури реєстру*, а не автоматичне підтвердження наукової істинності.


````python
edges = [(row.claim_id, sid)
         for row in claims.itertuples()
         for sid in row.source_ids]
missing = [(cid, sid) for cid, sid in edges if sid not in source_ids]
no_sources = claims.loc[claims["source_ids"].map(len).eq(0), "claim_id"].tolist()

print("Зв'язків claim-source:", len(edges))
print("Посилань на неіснуючі джерела:", missing)
print("Тверджень без джерел:", no_sources)
assert not missing

coverage = pd.DataFrame({
    "indicator": ["Частка тверджень із хоча б одним джерелом",
                  "Частка тверджень із валідними ідентифікаторами",
                  "Частка незалежно підтверджених тверджень"],
    "value": [claims["source_ids"].map(bool).mean(),
              claims["references_exist"].mean(),
              (claims["status"] == "supported").mean()]
})
display(coverage.round(3))
````

**Завдання 4.1.** Поясніть, чому показник структурного покриття не можна називати «доказовою достатністю». Запропонуйте додаткові поля реєстру для незалежної перевірки: DOI/URL, тип дослідження, метод, дата доступу, версія даних, оцінка ризику зміщення.

## 5. Перевірка узгодженості наукового твердження

Корисно розрізняти три рівні висновку:

1. *Описовий:* «У вибірці виявлено певну структуру».
2. *Прогнозний:* «За зафіксованого протоколу модель має визначену якість на цільовій вибірці».
3. *Причинний:* «За визначених припущень втручання змінює результат на оцінену величину».

Перехід між рівнями не відбувається автоматично. Зокрема, ні висока ROC AUC, ні стабільні кластери, ні SHAP-атрибуції не ідентифікують причинного ефекту [@rudin2019stop; @pearl2009causality].

## 6. Матриця загроз валідності

| Загроза | Контрольне питання | Можливе рішення |
|---|---|---|
| Витік даних | Чи є інформація з майбутнього або тестової частини? | Часовий/груповий split, аудит pipeline |

| Приховане змішування | Чи достатній набір коригування? | DAG, предметна аргументація, sensitivity analysis |
| Галюцинації LLM | Чи існує й чи підтримує цитоване джерело твердження? | Первинна перевірка посилань і тексту |
| Зсув розподілу | Чи поширюється висновок на іншу популяцію? | Зовнішня валідація |

Звіт має містити: мету й RQ; результати навчального експерименту (ранжування джерел, реєстр тверджень, аудит простежуваності); підсумкову таблицю артефактів v0.1–v0.7; одне основне наукове твердження з доказами та обмеженнями; реєстр перевірених джерел; структуру *Research Evidence Dossier v1.0*; висновки.

# Висновки

Завершальний етап ML/KDD-дослідження -- не отримання максимально складної моделі, а побудова *перевірюваного й відтворюваного переходу від даних до обґрунтованого наукового твердження*. Фундаментальні моделі та пошук можуть допомагати формувати гіпотези й знаходити джерела, але не замінюють наукової валідації [@fayyad1996kdd; @bommasani2021foundation].
