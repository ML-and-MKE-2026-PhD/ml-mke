## Мета лекції

Сформувати цілісне уявлення про машинне навчання (Machine Learning, ML) та вилучення знань з даних (Knowledge Discovery in Databases, KDD) не лише як сукупність алгоритмів, а як *інструментарій наукового пізнання*.

Після лекції здобувач має вміти:

- пояснювати історико-генетичний перехід від символічного ШІ до сучасних ML/KDD-парадигм;
- розрізняти *дані, інформацію, закономірність, модель, доказ і знання*;
- відрізняти прогнозну корисність моделі від наукової обґрунтованості твердження;
- визначати, які додаткові умови потрібні, щоб результат алгоритму міг підтримувати науковий висновок;
- позиціонувати власну дисертаційну задачу в ланцюгу ML/KDD-дослідження.

> Центральне питання лекції: **коли результат роботи алгоритму машинного навчання перестає бути просто числовим результатом і стає науковим свідченням?**

## 1. Чому для PhD недостатньо «побудувати модель»

У прикладному ML типовою метою може бути мінімізація помилки прогнозу. У дисертаційному дослідженні цього недостатньо. Висока метрика ще не відповідає на питання:

1. *що саме* було виявлено;
2. чи є результат стійким;
3. чи узагальнюється він за межі конкретної вибірки;
4. чи не породжений він способом підготовки даних;
5. чи допускає результат змістову інтерпретацію;
6. чи підтримує він причинне твердження;
7. які межі валідності має сформульований висновок.

Це принципово відрізняє **ML як інженерний інструмент** від **ML як інструменту наукового дослідження**. Класичне розрізнення алгоритмічного та пояснювального способів моделювання добре видно вже у дискусії Л. Бреймана про «дві культури» статистичного моделювання [@breiman2001two].

У цьому курсі нас цікавить не лише функція

$$
\hat{f}: X \rightarrow Y,
$$

а весь ланцюг переходу від спостережень до обґрунтованого твердження про досліджувану систему.

## 2. ML, Data Mining і KDD -- не синоніми

*Машинне навчання (Machine Learning, ML)* зосереджується на алгоритмах і моделях, здатних покращувати виконання певної задачі на основі досвіду/даних [@mitchell1997ml].

*Інтелектуальний аналіз даних (Data Mining)* орієнтований на виявлення нетривіальних структур, залежностей і патернів у даних.

*Вилучення знань з баз даних (Knowledge Discovery in Databases, KDD)* -- ширший процес, у якому власне Data Mining є лише одним з етапів. Класична постановка KDD підкреслює валідність, новизну, потенційну корисність і зрозумілість виявлених закономірностей [@fayyad1996kdd].

Отже, для нашого курсу корисно мислити так:

```{mermaid}
%%| fig-cap: "Співвідношення KDD, Data Mining і ML"
flowchart LR
    A("Дослідницька проблема") --> B("Дані")
    B --> C("Підготовка та представлення")
    C --> D("ML / Data Mining")
    D --> E("Виявлені закономірності")
    E --> F("Валідація та інтерпретація")
    F --> G("Наукові свідчення")
    G --> H("Обґрунтоване знання")
```

ML може бути центральним обчислювальним механізмом KDD, але **модель сама по собі не є знанням**.

## 3. Історико-генетична еволюція: що вважалося «знанням»

Історію штучного інтелекту доцільно розглядати не як послідовність модних алгоритмів, а як зміну способів *репрезентації та обґрунтування знання*.

| Парадигма | Що є центральним об'єктом | Типовий результат | Епістемічне питання |
|---|---|---|---|
| Символічний ШІ | правила, логічні конструкції | база правил, висновок | чи коректні правила і база знань? |
| Статистичне навчання | статистичні залежності | оцінена функція/параметри | чи узагальнюється залежність? |
| Data Mining / KDD | патерни в масивах даних | закономірність | чи є патерн валідним і корисним? |
| Глибоке навчання | багаторівневі представлення | learned representations | що саме закодовано у представленні? |
| XAI | поведінка моделі | пояснення прогнозу/моделі | чи пояснюємо ми модель або явище? |
| Causal ML | ефекти втручань | причинний ефект | які припущення дозволяють причинний висновок? |
| Foundation Models | універсальні навчені представлення | генерація, embeddings, адаптація | звідки походить і наскільки надійне «знання»? |

Статистична теорія навчання формалізувала проблему узагальнення [@vapnik1998statistical], глибоке навчання перенесло значну частину конструювання ознак у процес навчання [@lecun2015deep; @bengio2013representation], а фундаментальні моделі радикально розширили масштаб повторного використання навчених представлень [@bommasani2021foundation].

## 4. Від символічного ШІ до статистичного навчання

У символічній парадигмі знання переважно задавалося *явно*. Наприклад:

$$
(A \land B) \Rightarrow C.
$$

Якщо правила правильні, механізм логічного висновку дозволяє отримати нове твердження.

У статистичному та машинному навчанні логіка змінюється. Ми маємо вибірку

$$
D=\{(x_i,y_i)\}_{i=1}^{n},
$$

обираємо клас моделей $\mathcal{F}$ та шукаємо

$$
\hat f=\arg\min_{f\in\mathcal{F}}
\left[
\frac{1}{n}\sum_{i=1}^{n}L(y_i,f(x_i))+\lambda\Omega(f)
\right].
$$

Алгоритм знаходить функцію, яка добре поводиться за визначеним критерієм. Але з математичного факту мінімізації емпіричного ризику **не випливає автоматично**, що модель відкрила механізм реального процесу [@hastie2009elements; @bishop2006prml].

Це перший фундаментальний розрив:

> **прогнозна закономірність ≠ механістичне пояснення.**

## 5. KDD: закономірність ще потрібно перетворити на знання

Класична ідея KDD важлива для PhD саме тим, що вона не завершується запуском алгоритму [@fayyad1996kdd].

```{mermaid}
%%| fig-cap: "Спрощений дослідницький цикл KDD"
flowchart LR
    A("Проблема") --> B("Відбір даних")
    B --> C("Очищення")
    C --> D("Представлення")
    D --> E("Пошук закономірностей")
    E --> F("Оцінювання")
    F --> G("Інтерпретація")
    G --> H("Наукове твердження")
    H -. "перевірка / уточнення" .-> A
```

Особливо важливі дві обставини.

По-перше, результат залежить не лише від «реальності», а й від *ланцюга дослідницьких рішень*: вибірки, ознак, способу масштабування, метрики, моделі, гіперпараметрів і критерію валідації.

По-друге, KDD є *ітеративним*. Неочікуваний результат може змусити переглянути вихідну постановку, спосіб збору даних або навіть саме дослідницьке питання.

## 6. Ланцюг «Дані -- Інформація -- Закономірності -- Моделі -- Докази -- Знання»

Цей ланцюг є методологічною віссю всього курсу.

```{mermaid}
%%| fig-cap: "Від даних до наукового знання"
flowchart LR
    A("Дані") --> B("Інформація")
    B --> C("Закономірності")
    C --> D("Моделі")
    D --> E("Докази")
    E --> F("Знання")
```

Розглянемо переходи окремо.

**Дані → Інформація.** Спостереження набувають змісту лише після визначення семантики змінних, одиниць аналізу, процедури вимірювання та контексту.

**Інформація → Закономірності.** Алгоритм виявляє повторювану структуру. Але патерн може бути нестійким, випадковим або спричиненим способом представлення даних.

**Закономірності → Моделі.** Виявлена структура формалізується у вигляді прогнозної, описової, латентної або причинної моделі.

**Моделі → Докази.** Потрібні валідація, перевірка стійкості, оцінювання невизначеності, порівняння з альтернативами та аудит загроз валідності.

**Докази → Знання.** Лише після змістової інтерпретації та визначення меж застосовності результат може підтримувати наукове твердження.

Отже:

$$
\text{дані} \not\Rightarrow \text{знання}
$$

без проміжної системи методологічно обґрунтованих перетворень.

## 7. Невеликий обчислювальний експеримент: висока точність ще нічого не «пояснює»

Створимо синтетичні дані, де один предиктор є корисним для прогнозування. Навмисно почнемо з простої моделі. Мета прикладу -- не продемонструвати класифікатор, а побачити різницю між *метрикою* та *науковим висновком*.

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score

rng = np.random.default_rng(42)
n = 1000

x1 = rng.normal(size=n)
x2 = rng.normal(size=n)

# y залежить від x1; x2 тут є шумовою змінною.
p = 1 / (1 + np.exp(-(2.2 * x1)))
y = rng.binomial(1, p)

df = pd.DataFrame({"x1": x1, "x2": x2, "y": y})
df.head()
````

````python
X_train, X_test, y_train, y_test = train_test_split(
    df[["x1", "x2"]],
    df["y"],
    test_size=0.30,
    random_state=42,
    stratify=df["y"]
)

model = LogisticRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)
prob = model.predict_proba(X_test)[:, 1]

results = pd.Series({
    "Accuracy": accuracy_score(y_test, pred),
    "ROC AUC": roc_auc_score(y_test, prob),
    "coef_x1": model.coef_[0, 0],
    "coef_x2": model.coef_[0, 1]
})

results.round(3)
````

### Як правильно прочитати результат

Припустимо, ми отримали високу ROC AUC і великий коефіцієнт для $x_1$. Коректні твердження можуть бути такими:

- у *цьому експерименті* $x_1$ містить інформацію, корисну для прогнозування $y$;
- задана модель виявила стійку прогнозну асоціацію, якщо це підтверджено повторною валідацією.

Але з цього **не випливає**, що:

- $x_1$ є причиною $y$;
- зміна $x_1$ шляхом втручання змінить $y$;
- знайдений коефіцієнт описує механізм явища;
- результат автоматично переноситься на іншу популяцію.

Саме ця дисципліна переходу від *результату* до *твердження* буде повторюватися протягом усього курсу.

## 8. Навчені представлення: від ознак дослідника до ознак моделі

Класичний ML часто передбачає явне конструювання вектора ознак:

$$
x=(x_1,x_2,\ldots,x_p).
$$

У навчанні представлень (representation learning) модель сама формує відображення

$$
z=\phi_\theta(x),
$$

де $z$ -- нове представлення об'єкта, а параметри $\theta$ визначаються з даних [@bengio2013representation].

Це суттєва епістемологічна зміна. Якщо раніше дослідник значною мірою контролював семантику ознак, то тепер частина «мови опису» явища формується алгоритмом.

Звідси нове питання:

> Якщо модель виявила структуру в латентному просторі $z$, то що саме ми дізналися про досліджуваний об'єкт, а що -- лише про спосіб його представлення моделлю?

До цього питання повернемося у темах про валідацію прихованої структури та XAI.

## 9. Пояснення моделі і пояснення явища

Методи поясненного штучного інтелекту (Explainable Artificial Intelligence, XAI) можуть показувати, які ознаки вплинули на прогноз, як змінюється відповідь моделі або які локальні області простору визначають рішення.

Однак навіть якісне пояснення поведінки моделі не гарантує пояснення реального механізму. У високоризикових задачах це породжує окрему дискусію щодо переваги інтерпретованих моделей над постфактум-поясненнями [@rudin2019stop].

Фіксуємо принцип курсу:

$$
\boxed{\text{Пояснення моделі} \neq \text{Пояснення явища}}
$$

Аналогічно:

$$
\boxed{\text{Важливість ознаки} \neq \text{Причинна значущість}}
$$

## 10. Причинний поворот: від спостереження до втручання

Прогнозне машинне навчання переважно працює з розподілами на кшталт

$$
P(Y\mid X=x).
$$

Причинне питання має іншу природу: що станеться з $Y$, якщо *втрутитися* і встановити $X=x$? У термінах структурного причинного підходу це пов'язано з

$$
P(Y\mid do(X=x)).
$$

Загалом

$$
P(Y\mid X=x) \neq P(Y\mid do(X=x)).
$$

Різниця між спостереженням та втручанням є центральною для причинного висновку [@pearl2009causality; @hernan2020causal]. Сучасний Causal ML поєднує гнучкість ML з явними причинними припущеннями та процедурами оцінювання ефектів [@peters2017elements].

Тому «краще передбачає» і «краще відповідає на причинне питання» -- різні критерії.

## 11. Фундаментальні моделі: новий рівень автоматизації роботи зі знаннями

Фундаментальні моделі (foundation models) навчаються на великих і різнорідних масивах даних та адаптуються до широкого класу задач [@bommasani2021foundation]. Для дослідника вони можуть виступати як:

- джерело навчених представлень;
- інструмент пошуку і структурування інформації;
- генератор кандидатних гіпотез;
- засіб програмної та аналітичної підтримки;
- компонент складніших агентних дослідницьких систем.

Але генеративна здатність не тотожна епістемічній надійності. Текст, який *виглядає як знання*, може бути непідтвердженим.

```{mermaid}
%%| fig-cap: "Місце фундаментальної моделі у дослідницькому циклі"
flowchart LR
    A("Наукове питання") --> B("Джерела та дані")
    B --> C("Foundation model")
    C --> D("Представлення / гіпотези / узагальнення")
    D --> E("Незалежна перевірка")
    E --> F("Доказова база")
    F --> G("Наукове твердження")
```

Отже, фундаментальна модель може *підтримувати* процес вилучення знань, але не скасовує необхідності верифікації джерел, даних, обчислень і висновків.

## 12. Коли результат ML може бути науковим свідченням?

Запропонуємо робочу рамку, яку будемо деталізувати протягом курсу. Результат ML/KDD набуває доказової цінності, якщо дослідник може обґрунтувати щонайменше такі компоненти:

| Компонент | Контрольне питання |
|---|---|
| Постановка | Чи відповідає ML-задача науковому питанню? |
| Дані | Чи релевантні дані цільовій популяції/процесу? |
| Представлення | Чи не створює спосіб кодування штучний патерн? |
| Дизайн | Чи виключено витік та некоректне порівняння? |
| Узагальнення | На яку область можна переносити результат? |
| Невизначеність | Наскільки стабільна оцінка? |
| Робастність | Чи зберігається висновок за розумних альтернатив? |
| Інтерпретація | Чи не виходить твердження за межі результату? |
| Причинність | Якщо заявлена причинність -- чи ідентифікований ефект? |
| Відтворюваність | Чи може незалежний дослідник повторити процедуру? |

Жодна окрема метрика не замінює цю систему аргументації.

## 13. Три рівні тверджень: прогноз, пояснення, причинність

Розглянемо три формулювання.

**A. Прогнозне твердження**

> Модель на основі $X$ прогнозує $Y$ з визначеною якістю на незалежних даних.

**B. Інтерпретаційне твердження**

> Поведінка моделі систематично пов'язана з певними ознаками або структурами представлення.

**C. Причинне твердження**

> Втручання у $X$ змінює $Y$ на величину, оцінену за визначених ідентифікаційних припущень.

Ці твердження не утворюють автоматичної драбини

$$
A \not\Rightarrow B \not\Rightarrow C.
$$

Перехід між ними потребує *додаткових припущень, дизайну та доказів*.

Це одна з ключових ідей курсу: **тип наукового твердження має відповідати типу отриманого доказу**.

## 14. Наскрізна логіка курсу

Вісім тем утворюють не каталог методів, а послідовність дослідницьких питань.


Лабораторні роботи паралельно формуватимуть *наскрізний методологічний каркас власного дисертаційного ML/KDD-дослідження*.

## 15. Мінікейс для обговорення

Дослідник побудував модель виявлення певного типу поведінки користувачів соціальної платформи. На тестовій вибірці ROC AUC = 0.94. SHAP показує, що найважливішою ознакою є частота публікацій у нічний час.

Дослідник формулює висновок:

> «Нічна активність є ключовою причиною досліджуваної поведінки».

### Питання

1. Що фактично підтверджує ROC AUC = 0.94?
2. Що фактично показує SHAP-аналіз?
3. Які альтернативні пояснення можуть існувати?
4. Чи можна перейти від важливості ознаки до причинного твердження?
5. Які дані або дизайн дослідження були б потрібні для сильнішого висновку?

### Очікувана логіка відповіді

Метрика підтримує твердження про *прогнозну здатність у межах конкретної схеми оцінювання*. SHAP характеризує *поведінку моделі*, а не причинний механізм явища. Причинний висновок потребує окремої причинної постановки, припущень та ідентифікаційної стратегії.

## 16. Зв'язок із лабораторною роботою 1

Лабораторна робота 1 має назву:

> **«Формалізація дослідницької проблеми як задачі машинного навчання та вилучення знань»**.

Її завданням буде не «взяти датасет і навчити модель», а побудувати перший артефакт наскрізного дослідження:

$$
\boxed{
RQ
\rightarrow
ML/KDD\text{-задача}
\rightarrow
дані
\rightarrow
очікуваний\ тип\ результату
\rightarrow
допустиме\ наукове\ твердження
}
$$

Цей артефакт надалі буде уточнюватися в усіх лабораторних роботах курсу.

## 17. Підсумки

Після цієї лекції важливо зафіксувати п'ять тез.

1. **ML не тотожне KDD.** Алгоритм є частиною ширшого процесу переходу від даних до знань.
2. **Висока прогнозна якість не є достатнім науковим доказом.**
3. **Виявлена закономірність може залежати від даних, представлення, алгоритму та дизайну експерименту.**
4. **Пояснення моделі не тотожне поясненню явища, а асоціація не тотожна причинності.**
5. **Наукова цінність ML/KDD-результату визначається не лише метрикою, а якістю всього доказового ланцюга.**

Саме тому предметом цього курсу є не просто машинне навчання, а *методологічно контрольоване використання ML/KDD для отримання науково обґрунтованих тверджень*.
